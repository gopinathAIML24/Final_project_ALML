```python
import pandas as pd
import joblib

# Load datasets
train_df = pd.read_csv(
    r"C:\Users\GOPINATH\ALML\data\train.csv"
)

test_df = pd.read_csv(
    r"C:\Users\GOPINATH\ALML\data\test.csv"
)

# Load the SAME TF-IDF vectorizer from Day 2
vectorizer = joblib.load(
    r"C:\Users\GOPINATH\ALML\Models\tfidf_vectorizer.pkl"
)

# Prepare labels
y_train = train_df["category"]
y_test = test_df["category"]

# Transform text using the saved vectorizer
X_train_tfidf = vectorizer.transform(train_df["text"])
X_test_tfidf = vectorizer.transform(test_df["text"])

print("Training TF-IDF:", X_train_tfidf.shape)
print("Test TF-IDF:", X_test_tfidf.shape)
print("Training labels:", len(y_train))
print("Test labels:", len(y_test))
```

```python
from sklearn.svm import LinearSVC

svm_model = LinearSVC(
    C=1.0,
    max_iter=1000,
    dual="auto"
)

svm_model.fit(X_train_tfidf, y_train)

print("Linear SVM training completed!")
```

Day 3 — Step 3: SVM Predictions

```python
y_pred_svm = svm_model.predict(X_test_tfidf)

print("Number of SVM predictions:", len(y_pred_svm))
```

Step 4: Calculate SVM metrics

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

svm_accuracy = accuracy_score(y_test, y_pred_svm)

svm_precision = precision_score(
    y_test,
    y_pred_svm,
    average="weighted"
)

svm_recall = recall_score(
    y_test,
    y_pred_svm,
    average="weighted"
)

svm_f1 = f1_score(
    y_test,
    y_pred_svm,
    average="weighted"
)

print("SVM Accuracy :", svm_accuracy)
print("SVM Accuracy (%):", svm_accuracy * 100)
print("SVM Precision:", svm_precision)
print("SVM Recall   :", svm_recall)
print("SVM F1 Score :", svm_f1)
```

```python
y_pred_tuned = svm_tuned.predict(X_test_tuned)

print("Number of predictions:", len(y_pred_tuned))
```
